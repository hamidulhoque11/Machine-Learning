# Question 10 (10 Marks)
Handle missing values in the Titanic dataset by:
- Imputing missing Age values using SimpleImputer with mean strategy
- Imputing missing Embarked values using most frequent strategy
- Imputing missing Cabin values using constant strategy with "Missing"

Finally, check whether any missing values remain.


```python
X_train.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
import numpy as np
age_mean=SimpleImputer(missing_values=np.nan,strategy='mean') #ekhane chaile median diyeoo null value golo fill kora jay
age_mean.fit(X_train[['Age']])
X_train['Age']=age_mean.transform(X_train[['Age']])
X_test['Age']=age_mean.transform(X_test[['Age']])
# jeheto amra numerical feature er jonne missing value golo fill korchi tai strategy mean ba median use korte hocche
```

```python
sns.countplot(x=df['Embarked'])
plt.show()
```

```python
'''
ekhane categorycal feature er jonne missing value golo jeigolo ache
seigolo ke fill korte hole atfast countplot diye dekhbo je higest frequencey
pawa jaykina jodi pawa jay tahole highest frequency mode diye fill kore dibo
'''
embarked_imputer=SimpleImputer(missing_values=np.nan,strategy='most_frequent')
embarked_imputer.fit(X_train[['Embarked']])
X_train['Embarked']=embarked_imputer.transform(X_train[['Embarked']]).ravel()
X_test['Embarked']=embarked_imputer.transform(X_test[['Embarked']]).ravel()
```

```python
sns.countplot(x=df['Cabin'])
plt.show()
```

```python
'''
ar cabin oo same cagegorical features tai etar jonne oo dekhbo highest frequest mode value pawa jaykina ,
jeheto ekhane cabin er jonne frequencey buja jacche na tai missing string or indicator diye fill kore dichi
'''

cabin_imputer=SimpleImputer(missing_values=np.nan,strategy='constant',fill_value='Missing',add_indicator=True)
cabin_imputer.fit(X_train[['Cabin']])
X_train[['Cabin','cabin_indicator']]=cabin_imputer.transform(X_train[['Cabin']])
X_test[['Cabin','cabin_indicator']]=cabin_imputer.transform(X_test[['Cabin']])
```
