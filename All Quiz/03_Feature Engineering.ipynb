1. What is the main purpose of train-test split?

Ans: To evaluate model performance on unseen data

2. In a dataset of 1000 rows, if test_size=0.2, how many rows go to the training set?

Ans: 800

3. Which parameter ensures the same split every time you run train_test_split?

Ans : random_state

4. Which strategy is best for numerical data with outliers?

Ans : Median

5. Which strategy is used for categorical missing value imputation?

Ans : Most frequent

6. What does add_indicator=True in SimpleImputer do?

Ans: Adds a new column showing which values were missing

7. Why should you NOT fit the imputer on the test data?

Ans :  It causes data leakage

8. What should you do if a column has >70% missing values?

Ans : Drop the column

9. What happens if you forget .ravel() after transforming a single column with SimpleImputer?

Ans : Returns a 2D array and causes error when assigning to 1D column
