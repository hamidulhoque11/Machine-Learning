1. In a Countplot with Hue, what does the "hue" parameter do?

Ans : Adds a third categorical variable using color to differentiate groups

2. Which of the following is the correct use case for a Countplot?

Ans : Displaying the frequency of each category in a categorical variable

3. A Barplot for Cat-Num analysis is used to visualize which type of relationship?

Ans:One categorical and one numerical variable

4. What does a KDE (Kernel Density Estimation) plot show?

Ans :The smooth distribution curve of a continuous variable

5. Which plot would you use to see how many male and female passengers traveled in each ticket class on the Titanic?

Ans:Countplot with Hue

6. You want to compare the average salary of people from different cities. Which plot is most suitable?

Ans: Barplot

7. What does a Scatterplot primarily help us understand?

Ans : The relationship between two numerical variables

8. A student wants to understand the age distribution of customers. Which plot is best suited?

Ans: KDE plot

9. What is Y Data Profiling primarily used for in Exploratory Data Analysis (EDA)?

Ans : Getting a quick statistical summary and overview of the dataset

10. Which of the following plots is best for finding whether taller people tend to weigh more?

Ans : Scatterplot
