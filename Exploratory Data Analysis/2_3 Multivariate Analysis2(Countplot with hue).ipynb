# Categorical

## countplot(), groupby(), value_counts(normalize=True) 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('titanic_data_updated.csv')
df.head(3)
```

```python
sns.countplot(x=df['Pclass'],hue=df['Survived'])
plt.show()
```

```python
split_pclass=df.groupby('Pclass')['Survived'].value_counts()
split_pclass
```

```python
#persentage way
split_pclass=df.groupby('Pclass')['Survived'].value_counts(normalize=True)
split_pclass
```

```python
sns.countplot(x=df['Pclass'],hue=df['Sex'])
plt.show()
```

```python
sns.countplot(x=df['Pclass'],hue=df['Survived'],col=df['Sex'])
plt.show()
```

```python
#eta sikhayni nije theke try kore ber korchi  dekhar jonne aro category add korle kibabe kora jay
sns.catplot(x=df['Pclass'],hue=df['Survived'],col=df['Sex'],kind='count')
plt.show()
```
