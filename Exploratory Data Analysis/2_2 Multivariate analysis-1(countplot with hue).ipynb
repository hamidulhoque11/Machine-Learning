# Categorical

## countplot(), hue(), goroupby(), value_counts(normalized=True)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('titanic_data_updated.csv')
df.head(3)
```

```python
sns.countplot(x=df['Sex'],hue=df['Survived'])
plt.show()
```

Split gender

```python
# jodi split na kori tahole train,test korar somoy model bias hoye jabe cz male er frequencey beshi chilo

split_gender=df.groupby('Sex')['Survived'].value_counts()
print(split_gender)
```

Parsentage%

```python
split_gender=df.groupby('Sex')['Survived'].value_counts(normalize=True)
print(split_gender)

print(split_gender.index)


```

```python
split_gender=df.groupby('Sex')['Survived'].value_counts(normalize=True)# as parsentage
split_gender['male']
```
