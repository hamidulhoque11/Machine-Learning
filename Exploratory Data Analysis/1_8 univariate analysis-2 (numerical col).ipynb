## histplot(), kdeplot(), boxpot()

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('titanic_data_updated.csv')
df.sample(5)
```

```python
sns.histplot(df['Fare'],bins=50)
plt.title('Histogram for Fare')
plt.xlabel('Fare')
plt.ylabel('Frequencey')
plt.show()
```

```python
sns.kdeplot(df['Fare'])
plt.title('Histogram for Fare')
plt.xlabel('Fare')
plt.ylabel('Frequencey')
plt.show()
```

```python
sns.boxplot(df['Fare'])
plt.show()
```
