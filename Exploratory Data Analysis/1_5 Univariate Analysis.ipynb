```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv('titanic_data_updated.csv')
df.sample(3)
```

```python
sns.countplot(data=df,x=df['Survived'])
plt.show()
```

```python
#category wise ee show korbe kotojon survived hoiche ar koto jon korte pareni
df['Survived'].value_counts()
```

```python
(df['Survived'].value_counts()/len(df))*100 #% of servived
```

```python
survived_count=df['Survived'].value_counts()
labe=df['Survived'].unique()

plt.pie(survived_count,labels=labe)
```

```python
survived_count=df['Survived'].value_counts()
labe=df['Survived'].unique()

plt.pie(survived_count,labels=labe,autopct='%1.1f')
plt.title('pie chart of survival rate')
plt.show()
```
