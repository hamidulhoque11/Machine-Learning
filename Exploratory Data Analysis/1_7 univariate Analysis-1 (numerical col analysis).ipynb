## histplot(), kdeplot()

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('titanic_data_updated.csv')
df.sample(5)
```

```python
sns.histplot(df['Age'])
plt.show()
```

```python
#this is plot not a normal distrybution
#we have to use IQR mathod
sns.histplot(df['Age'],bins=50)
plt.title('Histogram of Age')
plt.xlabel('Age')
plt.ylabel('Frequencey')
plt.show()
```

```python
sns.kdeplot(df['Age'])
plt.title('KDE Plot of Age')
plt.xlabel('Age')
plt.ylabel('')
```
